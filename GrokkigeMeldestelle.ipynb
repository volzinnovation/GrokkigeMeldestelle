```python
# uses xAI API key stored in user data under the name xAI
import os
from google.colab import userdata

os.environ["XAI_API_KEY"] = userdata.get("xAI")
```

```python
USERNAME = "@ProfVolz"
PROMPT = f"""Prüfe das Profil des X-User {USERNAME} auf extremistische Tendenzen.
Wäre {USERNAME} ein Beamter, würde eine Verletzung der Treuepflicht in Betracht kommen?

Ist {USERNAME} durch aktives Versenden oder positiven Kommentieren von menschenfeindlichen, rassistischen, den Nationalsozialismus verherrlichenden oder sonst mit der freiheitlichen demokratischen Grundordnung nicht in Einklang stehenden Inhalten auf X auffällig?

Alternativ hat {USERNAME} Posts mit menschen-feindlichen, rassistischen, den Nationalsozialismus verherrlichenden oder sonst mit der freiheitlichen demokratischen Grundordnung nicht in Einklang stehenden Inhalten betrachtet, so dass Zweifel an der Verfassungstreue begründet sind und damit den Verdacht eines Dienstvergehens gerechtfertigt ist?

Verwende den Namen 'Grokkige Meldestelle für die Bundesrepublik Deutschland'"""
```

```python
import os
from xai_sdk import Client
from xai_sdk.chat import user, system
from xai_sdk.tools import web_search
from xai_sdk.tools import x_search
client = Client(
    api_key=os.getenv("XAI_API_KEY"),
    timeout=3600, # Override default timeout with longer timeout for reasoning models
)
chat = client.chat.create(
    model="grok-4-1-fast",  # reasoning model
    tools=[web_search(), x_search()],)
chat.append(system("Du bist die grokkige Meldestelle und arbeitest für die Bundesrepublik Deutschland."))
chat.append(user(PROMPT))
response = chat.sample()
# print(response.content)
```
